# Lab Task 2 — Single Forward Pass and Error

**Source:** *Understanding Deep Learning*, Laboratory Task 2.

**Instruction:** Perform a single forward pass using the given input, hidden and output weights, biases, and ReLU activation. Compute the prediction error.

Given

\[
x=\begin{bmatrix}1\\0\\1\end{bmatrix},\qquad y=1,\qquad f(z)=\max(0,z).
\]

The input-to-hidden weight matrix is

\[
W_h=\begin{bmatrix}0.2&-0.3\\0.4&0.1\\-0.5&0.2\end{bmatrix},\quad
b_h=\begin{bmatrix}-0.4&0.2\end{bmatrix}.
\]

The hidden-to-output weights are \(W_o=[-0.3,-0.2]^T\) and the output bias is \(b_o=0.1\).

**Calculation:**

\[
\begin{aligned}
z_1 &= 1(0.2)+0(0.4)+1(-0.5)-0.4=-0.7,\\
z_2 &= 1(-0.3)+0(0.1)+1(0.2)+0.2=0.1,\\
h &= [\operatorname{ReLU}(-0.7),\operatorname{ReLU}(0.1)]=[0,0.1],\\
z_o &=0(-0.3)+0.1(-0.2)+0.1=0.08,\\
\hat y&=\operatorname{ReLU}(0.08)=0.08.
\end{aligned}
\]

The signed error \(y-\hat y=0.92\); the squared error \((y-\hat y)^2=0.8464\). This is also the MSE for one scalar prediction.


```python
import numpy as np

x = np.array([1.0, 0.0, 1.0])
y = 1.0
W_h = np.array([[0.2, -0.3], [0.4, 0.1], [-0.5, 0.2]])
b_h = np.array([-0.4, 0.2])
W_o = np.array([-0.3, -0.2])
b_o = 0.1

def relu(z):
    return np.maximum(0, z)

z_h = x @ W_h + b_h
h = relu(z_h)
z_o = float(h @ W_o + b_o)
y_hat = float(relu(z_o))
error = y - y_hat
squared_error = error ** 2
half_squared_error = 0.5 * squared_error

print('Hidden weighted sums:', z_h)
print('Hidden activations:', h)
print(f'Output weighted sum: {z_o:.4f}')
print(f'Predicted output: {y_hat:.4f}')
print(f'Error (actual - predicted): {error:.4f}')
print(f'Squared error / one-sample MSE: {squared_error:.4f}')
print(f'Half-squared error: {half_squared_error:.4f}')

assert np.allclose(z_h, [-0.7, 0.1])
assert np.isclose(y_hat, 0.08)
assert np.isclose(squared_error, 0.8464)
```

**Interpretation:** The first hidden neuron is inactive because its weighted sum is negative. The model predicts **0.08** instead of the target **1.00**, so the initial prediction has a relatively large error. The next activity uses backpropagation to update the parameters.
