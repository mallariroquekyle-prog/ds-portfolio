# Lab Task 5 — NumPy and PyTorch Exercises

**Source:** *PyTorch Basics*, Laboratory Task 5.

This notebook answers **all ten** steps of the task, from imports and random seeds to matrix multiplication. Reproducibility is checked with the same seed as the provided lecture notebook.


```python
# Step 1. Perform standard imports
import numpy as np
import torch

# Step 2. Set seed without returning a value
def set_seed(seed: int) -> None:
    np.random.seed(seed)
    torch.manual_seed(seed)

# Step 3. Generate 6 NumPy integers between 0 (inclusive) and 5 (exclusive)
set_seed(42)
arr = np.random.randint(0, 5, size=6).astype(np.int32)
print('3. NumPy array:', arr)

# Step 4. Create tensor from array
x = torch.from_numpy(arr)
print('4. Tensor:', x)

# Step 5. Convert int32 to int64
x = x.to(torch.int64)
print('5. dtype:', x.dtype)

# Step 6. Reshape to 3x2
x = x.reshape(3, 2)
print('6. Reshaped x:\n', x)

# Step 7. Get right-hand column
right_column = x[:, 1]   # one-dimensional column, containing 3 entries
print('7. Right column:', right_column)

# Step 8. Square x values WITHOUT changing x
x_squared = x.pow(2)
print('8. Squared values:\n', x_squared)
print('   Original x remains:\n', x)

# Step 9. Use PyTorch to create a compatible 2x3 matrix y
set_seed(42)
y = torch.randint(0, 5, (2, 3), dtype=torch.int64)
print('9. y:\n', y)

# Step 10. Matrix multiply: (3x2) @ (2x3) = (3x3)
product = x @ y
print('10. Matrix product:\n', product)

assert np.array_equal(arr, np.array([3, 4, 2, 4, 4, 1]))
assert x.dtype == torch.int64 and x.shape == (3, 2)
assert torch.equal(right_column, torch.tensor([4, 4, 1]))
assert torch.equal(x_squared, torch.tensor([[9, 16], [4, 16], [16, 1]]))
assert torch.equal(y, torch.tensor([[2, 2, 1], [4, 1, 0]]))
assert torch.equal(product, torch.tensor([[22, 10, 3], [20, 8, 2], [12, 9, 4]]))
print('All ten tasks passed the checks.')
```

**Explanation:** After reshaping, \(x\) has shape \((3,2)\). To multiply it, \(y\) must have shape \((2,3)\), which produces a \((3,3)\) output. `x.pow(2)` returns a new tensor and does not modify `x`.
